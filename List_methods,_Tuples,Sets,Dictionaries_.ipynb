## List methods

**index(element): returns the position of the element in the list. If the element is not present, it will throw an error**

**count(element) : returns the number of occurrences of that element in the list. If it is not present, it will return 0**

**Add elements to the list: At a time we can add only one element to the list.**

1) append(element): adds the element at the end of the list
2) insert(position,element): adds the element at the specified position in the list.

**Remove the elements from the list.At a time we can remove only one element to the list.**

1) remove(element) : removes the element from the list. If the element is not present,it will throw an error
2) pop(position) : removes the element present at that position. If position is not passed i.e pop() will always remove the last element from the list
3) discard(element) : removes the element from the list. If the element is not present,it will NOT throw an error

**You can also delete single/multiple elements using del()**

    example: del(list_variable) : deletes list_vairable permanently.
    del(list_variable[2:5]) : deletes the elements present in the list_variable from position 2 upto 4.

**Merge different lists together into a single list**

1) list1.extend(list2) : list2 gets merged into list1
2) new_list =list1 + list2  : you can also use + operator between 2 lists to merge them. But assign it to a new variable to access it

## Dictionary

1) Allows to store data in key-value pairs
2) key must be unique in nature. key can be integer,string or float datatype
3) values can be duplicated in nature. values can be integers/float/strings/lists/tuples/sets
4) Dicationaries are Mutable in nature
5) Syntax: dict_name = {key1:value2,key2:value2,key3:value3}

**Access Dictionary elements:**

1) keys(): returns keys present in the dictionary
2) values() : returns values present in the dictionary
3) items() : returns key value pairs present in the dictionary

### Accessing dictionary elements
1) dict_name[key] : will return the corresponding value. If the key is not present, it will throw an error
2) dict_name.get(key) : will return the corresponding value. If the key is not present, it will return None

### Adding new key-value pair to the dictionary

dict_name[new_key] = new_value

### Updating existing dictionary key value pair

dict_name[key] = new_value

### Removing key-value pairs

1) pop(key) : removes the key value pair on basis of the key given
2) popitem() : removes the last inserted key-value pair.

**del(dict_name[key])**: Removes the specific key-value pair.

**del(dict_name)**: Deletes dictionary permanently
